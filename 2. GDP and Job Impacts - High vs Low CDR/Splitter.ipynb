```python
import pandas as pd

# Read the Excel file
df_GDP = pd.read_csv("GDP by sector (CAD 2015).csv")
df_jobs = pd.read_csv("jobs by sector (thousand FTEs).csv")

# Define the columns to be selected
columns_to_select = ['year', 'policy', 'swb', 'ccs', 'hyd', 'oil', 'lul']

# Split the data based on the 'dac' column for GDP
yes_data_GDP = df_GDP[df_GDP['dac'] == 'yes'][columns_to_select + ['dac', 'agriculture and forestry', 'oil and gas', 'mining', 'utilities', 'construction', 'manufacturing', 'transportation', 'services', 'direct air capture']]
no_data_GDP = df_GDP[df_GDP['dac'] == 'no'][columns_to_select + ['dac', 'agriculture and forestry', 'oil and gas', 'mining', 'utilities', 'construction', 'manufacturing', 'transportation', 'services', 'direct air capture']]

# Split the data based on the 'dac' column for jobs
yes_data_jobs = df_jobs[df_jobs['dac'] == 'yes'][columns_to_select + ['dac', 'agriculture and forestry', 'oil and gas', 'mining', 'utilities', 'construction', 'manufacturing', 'transportation', 'services', 'direct air capture']]
no_data_jobs = df_jobs[df_jobs['dac'] == 'no'][columns_to_select + ['dac', 'agriculture and forestry', 'oil and gas', 'mining', 'utilities', 'construction', 'manufacturing', 'transportation', 'services', 'direct air capture']]

# Sort the dataframes based on columns x1 to x6
yes_data_GDP = yes_data_GDP.sort_values(by=columns_to_select)
no_data_GDP = no_data_GDP.sort_values(by=columns_to_select)

# Sort the dataframes based on columns x1 to x6
yes_data_jobs = yes_data_jobs.sort_values(by=columns_to_select)
no_data_jobs = no_data_jobs.sort_values(by=columns_to_select)

# Save the divided data to new Excel files or process them as needed
yes_data_GDP.to_excel("yes_data_GDP.xlsx", index=False)
no_data_GDP.to_excel("no_data_GDP.xlsx", index=False)

# Save the divided data to new Excel files or process them as needed
yes_data_jobs.to_excel("yes_data_jobs.xlsx", index=False)
no_data_jobs.to_excel("no_data_jobs.xlsx", index=False)
```
