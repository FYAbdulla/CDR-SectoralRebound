```python
"""
Created on Tue Aug 20 21:17:41 2024

@author: fyabdulla
"""

import matplotlib.pyplot as plt
import numpy as np

# Input the data manually
years = [2040, 2045, 2050]

ldt_min = [0.008152971, 0.013724116, 0.038630248]
ldt_max = [0.053982102, 0.074018026, 0.150810734]
ldt_med = [0.027335262, 0.040115395, 0.077486765]

mhdt_min = [0.19043557, 0.134156333, 0.135153222]
mhdt_max = [0.429181757, 0.26080487, 0.235255515]
mhdt_med = [0.272860748, 0.1916498, 0.186243712]

og_min = [0.108638311, 0.111718386, 0.07665668]
og_max = [0.375532514, 0.329446938, 0.232523657]
og_med = [0.190501871, 0.154788678, 0.116137141]

industry_min = [0.167757056, 0.220364901, 0.21109693]
industry_max = [0.334239791, 0.377527063, 0.347389784]
industry_med = [0.258970656, 0.275212186, 0.276770027]

buildings_min = [0.035915263, 0.077554618, 0.117126032]
buildings_max = [0.177863231, 0.16005712, 0.175477579]
buildings_med = [0.129574102, 0.127050363, 0.142558177]

# Define categories
categories = [
    ('Light Transport', ldt_min, ldt_max, ldt_med, 'red'),
    ('Med/Hvy Transport', mhdt_min, mhdt_max, mhdt_med, 'blue'),
    ('Oil & Gas', og_min, og_max, og_med, 'green'),
    ('Industry', industry_min, industry_max, industry_med, 'goldenrod'),
    ('Buildings', buildings_min, buildings_max, buildings_med, 'black'),
]

# Create one row of 5 subplots with shared y-axis
fig, axes = plt.subplots(
    nrows=1,
    ncols=5,
    figsize=(22, 6),
    sharey=True
)

for ax, (category, min_values, max_values, med_values, color) in zip(axes, categories):

    min_values = np.array(min_values, dtype=np.float64)
    max_values = np.array(max_values, dtype=np.float64)
    med_values = np.array(med_values, dtype=np.float64)

    # Plot min/max bounds
    ax.plot(years, min_values, linestyle='-', color=color, linewidth=2)
    ax.plot(years, max_values, linestyle='-', color=color, linewidth=2)

    # Fill between min and max
    ax.fill_between(years, min_values, max_values, color=color, alpha=0.2)

    # Plot median
    ax.plot(years, med_values, linestyle='--', color=color, linewidth=2.5)

    # Subplot title
    ax.set_title(category, fontsize=16, fontweight='bold')

    # X-axis formatting
    ax.set_xticks(years)
    ax.set_xticklabels(years, fontsize=13, fontweight='bold')

    # Grid
    ax.grid(True, linestyle='--', alpha=0.35)

    # Annotate 2050 max and min
    x_2050 = years[-1]
    y_min_2050 = min_values[-1]
    y_max_2050 = max_values[-1]

    ax.annotate(
        f"Max: {y_max_2050:.2f}",
        xy=(x_2050, y_max_2050),
        xytext=(-45, 12),
        textcoords="offset points",
        fontsize=11,
        fontweight='bold',
        arrowprops=dict(arrowstyle='->', lw=1)
    )

    ax.annotate(
        f"Min: {y_min_2050:.2f}",
        xy=(x_2050, y_min_2050),
        xytext=(-45, -22),
        textcoords="offset points",
        fontsize=11,
        fontweight='bold',
        arrowprops=dict(arrowstyle='->', lw=1)
    )

# Shared y-axis label
axes[0].set_ylabel(
    'Emissions Rebound\n(Mt CO$_2$ eq / Mt Added CDR Capacity)',
    fontsize=15,
    fontweight='bold'
)

# Y-axis tick formatting
axes[0].tick_params(axis='y', labelsize=13)

# Optional: set same y-axis limits across all panels
axes[0].set_ylim(0, 0.45)

# Add a common legend manually
handles = [
    plt.Line2D([0], [0], color='black', linestyle='-', linewidth=2, label='Min/Max Range'),
    plt.Line2D([0], [0], color='black', linestyle='--', linewidth=2.5, label='Median'),
]

fig.legend(
    handles=handles,
    loc='lower center',
    ncol=2,
    fontsize=13,
    frameon=False,
    bbox_to_anchor=(0.5, -0.03)
)

# Layout adjustment
plt.tight_layout()
plt.subplots_adjust(bottom=0.18, wspace=0.08)

plt.show()
```
