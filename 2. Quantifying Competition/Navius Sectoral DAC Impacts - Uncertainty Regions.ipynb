```python
"""
Created on Tue Aug 20 21:17:41 2024

@author: fyabdulla
"""

import matplotlib.pyplot as plt
import numpy as np

# Input the data manually
years = [2040, 2045, 2050]

residential_min = [0.035, 0.070809821, 0.097936182]
residential_max = [0.073672845, 0.06942885, 0.073851937]

commercial_min = [0.030935597, 0.054854768, 0.060107135]
commercial_max = [0.060556534, 0.051081474, 0.050807186]

ldt_min = [0.008655463, 0.016350605, 0.042754505]
ldt_max = [0.049811714, 0.063950958, 0.13950869]

mhdt_min = [0.231491877, 0.146072847, 0.170729616]
mhdt_max = [0.260851864, 0.21921217, 0.185012211]

ot_min = [0.005975, 0.01274233, 0.024029765]
ot_max = [0.006963092, 0.012381448, 0.018143812]

og_min = [0.188725361, 0.159861424, 0.101920401]
og_max = [0.308057077, 0.267489965, 0.195381308]

industry_min = [0.226065361, 0.257075125, 0.276507608]
industry_max = [0.282138199, 0.278166129, 0.273968878]

aw_min = [0.044643893, 0.061014551, 0.072033998]
aw_max = [0.050195663, 0.06819618, 0.064]

utilities_min = [0.017613838, 0.055054349, 0.051504073]
utilities_max = [0.136354217, 0.153240005, 0.128756368]

# Define categories and their colors for consistent plotting
categories = [
    ('Residential Buildings', residential_min, residential_max, 'blue'),
    ('Commercial Buidlings', commercial_min, commercial_max, 'orange'),
    ('Light Transport', ldt_min, ldt_max, 'orange'),
    ('Med/Hvy Transport', mhdt_min, mhdt_max, 'red'),
    ('Other Transport', ot_min, ot_max, 'purple'),
    ('Oil & Gas', og_min, og_max, 'green'),
    ('Industry', industry_min, industry_max, 'blue'),
    ('Agriculture & Waste', aw_min, aw_max, 'cyan'),
    ('Utilities', utilities_min, utilities_max, '#708090'),
]

# Plot each category
plt.figure(figsize=(12, 8))

for category, min_values, max_values, color in categories:
    min_values = np.array(min_values, dtype=np.float64)
    max_values = np.array(max_values, dtype=np.float64)
    
    plt.plot(years, min_values, label=f'{category}', linestyle='-', color=color)
    plt.plot(years, max_values, label=None, linestyle='-', color=color)
    plt.fill_between(years, min_values, max_values, color=color, alpha=0.2)

# Set x-ticks to only show specific years
plt.xticks([2040, 2045, 2050], fontsize=19, fontweight='bold')
plt.yticks(fontsize=19)

# Add labels and title
#plt.xlabel('Year', fontsize=20)
plt.ylabel('Emission Differentials (Mt CO$_2$ eq / Mt CO$_2$ Sequestered)', fontsize=19, fontweight='bold')
#plt.title('Navius gTech-IESD: Sectoral Impact of DAC Deployment', fontsize=23, fontweight='bold')
plt.legend(loc='upper left', bbox_to_anchor=(1.01, 1), ncol=1, fontsize=19)
plt.subplots_adjust(bottom=-0.16)

# Display the plot
plt.show()

```
