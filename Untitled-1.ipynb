```python
import matplotlib.pyplot as plt

# Data for the bar chart
categories = ['Transportation', 'Industry', 'Commercial', 'Residential', 'Electricity']
values = [0.09, 0.01, 0.02, 0.035, -0.35]

colors = ['blue', 'purple', 'orange', 'red', 'yellow']

# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(categories, values, color=colors, edgecolor='black')

# Add gridlines
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add title and labels
plt.title("Emission Differentials per Unit of Added Sequestration Capacity - Japan - 2050", fontsize=14)
plt.xlabel("Sectors", fontsize=12)
plt.ylabel("Differential (Units)", fontsize=12)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data for the line chart (manually input values from 2020 onwards)
years = [2020, 2025, 2030, 2035, 2040, 2045, 2050]
sequestration_values = [5, 7, 10, 15, 22, 35, 50]  # Hypothetical values based on observations

# Create the line chart
plt.figure(figsize=(10, 6))
plt.plot(years, sequestration_values, marker='o', label='Total Sequestration')

# Add title and labels
plt.title("Total Sequestration (Summed Averages) - Japan", fontsize=14)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Sequestration Magnitude (Mt CO2/yr)", fontsize=12)

# Add gridlines
plt.grid(axis='both', linestyle='--', alpha=0.7)

# Add legend
plt.legend()

# Display the chart
plt.tight_layout()
plt.show()
```
