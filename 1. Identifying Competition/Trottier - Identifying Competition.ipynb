```python
import numpy as np
import matplotlib.pyplot as plt

# Data extracted from Trottier's Canada Energy Outlook 2021
sectors = ["Fugitive Sources", "Energy Production", "Electricity", "Transport", "Industry- Process", "DAC"]
emissions_differentials = [2, 4, 5, 1, 1, -13]

# Create bar chart
plt.figure(figsize=(8, 6))
bars = plt.bar(sectors, emissions_differentials, color=['red', 'green', 'purple', 'orange', 'brown', 'blue'])

# Labels
plt.ylabel("Emissions Differentials (MT CO₂ eq)", fontweight='bold')

# Rotate and wrap x-axis labels for better readability
wrapped_labels = [label.replace(' ', '\n') for label in sectors]
plt.xticks(ticks=[0, 1, 2, 3, 4, 5], labels=wrapped_labels)
plt.yticks(np.arange(-14, 6 + 1, 2))
plt.grid(axis='y', alpha=0.3)

# Add data labels above or below each bar
for bar in bars:
    height = bar.get_height()
    if height >= 0:
        y_pos = height + 0.1
        va = 'bottom'
    else:
        y_pos = height - 0.1
        va = 'top'
    
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        y_pos,
        f'{height}',
        ha='center',
        va=va,
        fontsize=10,
        fontweight='bold'
    )

# Display plot
plt.show()

```
