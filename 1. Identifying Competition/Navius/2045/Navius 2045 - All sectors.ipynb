```python
"""
Created on Mon Mar 25 14:38:32 2024

@author: Feras Abdulla
"""

import matplotlib.pyplot as plt
import pandas as pd

# Read data from Excel file
df = pd.read_excel("Data for BW Plot.xlsx")

# Create boxplot
plt.figure(figsize=(24, 12))  # Adjust figure size as needed
plt.boxplot(df.values, labels=df.columns)

# Add labels for max, min
for i, col in enumerate(df.columns):
    plt.text(i + 1, df[col].max(), f'{df[col].max():.1f}', ha='center', va='bottom', color='red', fontsize=12)
    plt.text(i + 1, df[col].min(), f'{df[col].min():.1f}', ha='center', va='top', color='red', fontsize=12)

# Add labels for median to the right of each box
for i, col in enumerate(df.columns):
    plt.text(i + 1.25, df[col].median(), f'{df[col].median():.1f}', ha='left', va='center', color='red', fontsize=12)

# Add labels and title
plt.xlabel('Sectors', fontsize=16)
plt.ylabel('Emissions Deltas (Megatonnes)', fontsize=16)
plt.title('2045 Emissions Comparison - DAC On vs Off', fontsize=18)

# Rotate x-axis labels for better readability
plt.xticks(rotation=60)

# Show plot
plt.show()
```
